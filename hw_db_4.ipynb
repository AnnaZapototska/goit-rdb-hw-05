```python
customer_features_sql = """
WITH
-- Stage 1: Establish a stable reference date and calculate order totals
reference_date AS (
    SELECT MAX(order_purchase_timestamp)::DATE + 1 AS ref_date
    FROM olist_orders
),

stage1_orders_total AS (
    SELECT
        o.order_id,
        o.customer_id,
        o.order_status,
        o.order_purchase_timestamp,
        COALESCE(
            SUM(COALESCE(oi.price, 0) + COALESCE(oi.freight_value, 0)),
            0
        ) AS order_total
    FROM olist_orders o
    LEFT JOIN olist_order_items oi
        ON o.order_id = oi.order_id
    GROUP BY
        o.order_id,
        o.customer_id,
        o.order_status,
        o.order_purchase_timestamp
),

-- Stage 2: Calculate RFM features for each customer
stage2_rfm AS (
    SELECT
        ot.customer_id,
        (rd.ref_date - MAX(ot.order_purchase_timestamp)::DATE)
            AS recency_days,
        COUNT(DISTINCT ot.order_id) AS frequency,
        SUM(ot.order_total) AS monetary
    FROM stage1_orders_total ot
    CROSS JOIN reference_date rd
    GROUP BY ot.customer_id, rd.ref_date
),

-- Stage 3: Calculate the customer's cohort quarter
stage3_cohort AS (
    SELECT
        customer_id,
        DATE_TRUNC(
            'quarter',
            MIN(order_purchase_timestamp)
        )::DATE AS cohort_quarter
    FROM olist_orders
    GROUP BY customer_id
),

-- Stage 4: Calculate rolling 30-day average order value
stage4_rolling_aov AS (
    SELECT
        customer_id,
        order_id,
        AVG(order_total) OVER (
            PARTITION BY customer_id
            ORDER BY order_purchase_timestamp
            RANGE BETWEEN INTERVAL '30 days' PRECEDING
                      AND CURRENT ROW
        ) AS avg_order_value_30d,
        ROW_NUMBER() OVER (
            PARTITION BY customer_id
            ORDER BY order_purchase_timestamp DESC, order_id DESC
        ) AS rn
    FROM stage1_orders_total
),

-- Stage 5: Calculate the cancellation rate for each customer
stage5_cancel_rate AS (
    SELECT
        customer_id,
        COUNT(*) FILTER (
            WHERE order_status = 'canceled'
        )::NUMERIC / NULLIF(COUNT(*), 0) AS cancellation_rate
    FROM olist_orders
    GROUP BY customer_id
),

-- Stage 6: Calculate revenue by product category for each customer
stage6_category_revenue AS (
    SELECT
        o.customer_id,
        p.product_category_name,
        SUM(oi.price) AS category_revenue
    FROM olist_orders o
    JOIN olist_order_items oi
        ON o.order_id = oi.order_id
    JOIN olist_products p
        ON oi.product_id = p.product_id
    WHERE p.product_category_name IS NOT NULL
    GROUP BY
        o.customer_id,
        p.product_category_name
),

-- Stage 7: Select the customer's top category by revenue
stage7_preferred_category AS (
    SELECT
        customer_id,
        product_category_name AS preferred_category
    FROM (
        SELECT
            customer_id,
            product_category_name,
            ROW_NUMBER() OVER (
                PARTITION BY customer_id
                ORDER BY category_revenue DESC,
                         product_category_name ASC
            ) AS rn
        FROM stage6_category_revenue
    ) ranked_categories
    WHERE rn = 1
)

-- Final stage: Combine all customer-level features
SELECT
    r.customer_id,
    r.recency_days,
    r.frequency,
    r.monetary,
    c.cohort_quarter,
    a.avg_order_value_30d,
    cr.cancellation_rate,
    pc.preferred_category
FROM stage2_rfm r
LEFT JOIN stage3_cohort c
    ON r.customer_id = c.customer_id
LEFT JOIN stage4_rolling_aov a
    ON r.customer_id = a.customer_id
   AND a.rn = 1
LEFT JOIN stage5_cancel_rate cr
    ON r.customer_id = cr.customer_id
LEFT JOIN stage7_preferred_category pc
    ON r.customer_id = pc.customer_id
"""

with engine.connect() as conn:
    features_sample = pd.read_sql_query(
        text(customer_features_sql + " LIMIT 10"),
        conn
    )

display(features_sample)
print("Sample rows:", len(features_sample))
```

```python
create_referrals_sql = """
CREATE TABLE hw5_referrals (
    referrer_customer_id TEXT NOT NULL
        REFERENCES olist_customers(customer_id),
    referred_customer_id TEXT PRIMARY KEY
        REFERENCES olist_customers(customer_id),
    referred_at TIMESTAMP NOT NULL,
    CHECK (referrer_customer_id <> referred_customer_id)
);
"""

with engine.begin() as conn:
    conn.exec_driver_sql(create_referrals_sql)

print("hw5_referrals created successfully.")
```

```python
insert_referrals_sql = """
WITH ordered_customers AS (
    SELECT
        customer_id,
        ROW_NUMBER() OVER (ORDER BY customer_id) AS rn
    FROM olist_customers
    ORDER BY customer_id
    LIMIT 10
),
referral_chain AS (
    SELECT
        current_customer.customer_id AS referrer_customer_id,
        next_customer.customer_id AS referred_customer_id,
        next_customer.rn
    FROM ordered_customers current_customer
    JOIN ordered_customers next_customer
        ON next_customer.rn = current_customer.rn + 1
)
INSERT INTO hw5_referrals (
    referrer_customer_id,
    referred_customer_id,
    referred_at
)
SELECT
    referrer_customer_id,
    referred_customer_id,
    TIMESTAMP '2018-01-01'
        + rn * INTERVAL '1 day'
FROM referral_chain
ON CONFLICT DO NOTHING;
"""

with engine.begin() as conn:
    conn.exec_driver_sql(insert_referrals_sql)

with engine.connect() as conn:
    count = conn.execute(
        text("SELECT COUNT(*) FROM hw5_referrals")
    ).scalar_one()

print("Referral records:", count)
```

```python
recursive_sql = """
WITH RECURSIVE referral_tree AS (
    -- Anchor: find customers who were not referred by anyone
    SELECT
        c.customer_id,
        0 AS depth,
        ARRAY[c.customer_id] AS path_ids
    FROM olist_customers c
    WHERE NOT EXISTS (
        SELECT 1
        FROM hw5_referrals r
        WHERE r.referred_customer_id = c.customer_id
    )

    UNION ALL

    -- Recursive step: follow each referral to the next customer
    SELECT
        r.referred_customer_id AS customer_id,
        rt.depth + 1 AS depth,
        rt.path_ids || r.referred_customer_id AS path_ids
    FROM referral_tree rt
    JOIN hw5_referrals r
        ON r.referrer_customer_id = rt.customer_id
    WHERE NOT (r.referred_customer_id = ANY(rt.path_ids))
      AND rt.depth < 50
)
SELECT
    customer_id,
    depth,
    path_ids
FROM referral_tree
WHERE depth > 0
ORDER BY depth, customer_id
LIMIT 20;
"""

with engine.connect() as conn:
    referral_result = pd.read_sql_query(
        text(recursive_sql),
        conn
    )

display(referral_result)
```

```python
cycle_test_sql = """
WITH RECURSIVE
edges(referrer, referred) AS (
    VALUES
        ('X', 'A'),
        ('A', 'B'),
        ('B', 'A')
),
nodes AS (
    SELECT referrer AS node FROM edges
    UNION
    SELECT referred AS node FROM edges
),
referral_tree AS (
    -- Anchor: start from a node with no incoming referral
    SELECT
        n.node,
        0 AS depth,
        ARRAY[n.node] AS path_ids
    FROM nodes n
    WHERE NOT EXISTS (
        SELECT 1
        FROM edges e
        WHERE e.referred = n.node
    )

    UNION ALL

    -- Recursive step: follow edges while preventing cycles
    SELECT
        e.referred,
        rt.depth + 1,
        rt.path_ids || e.referred
    FROM referral_tree rt
    JOIN edges e
        ON e.referrer = rt.node
    WHERE NOT (e.referred = ANY(rt.path_ids))
      AND rt.depth < 50
)
SELECT node, depth, path_ids
FROM referral_tree
ORDER BY depth;
"""

with engine.connect() as conn:
    cycle_result = pd.read_sql_query(
        text(cycle_test_sql),
        conn
    )

display(cycle_result)
```

## Reflection

**1. Which feature calculation is the most expensive, and why?**

The execution plan shows that the pipeline is computationally expensive because it processes and aggregates data across 99,441 orders and 112,650 order items. The RFM aggregation and sorting take substantial time: the `GroupAggregate` stage runs from approximately 573 ms to 744 ms. The preferred-category calculation also requires joins, grouping by customer and product category, and ranking categories by revenue. Several sorting operations use external merge with disk, including sorts that use approximately 9.3 MB and 9.2 MB of temporary disk space. This indicates that intermediate operations consume additional resources.

**2. Should these features be cached in a Materialized View?**

Yes, when the features are queried frequently and do not need to be recalculated for every request. The full CTE pipeline took approximately 2,715.5 ms. In comparison, retrieving one customer from the Materialized View using its unique index took 0.042 ms. These queries return different amounts of data, so this is not a direct like-for-like benchmark, but it demonstrates the benefit of precomputing features for individual lookups.

**3. When is a regular VIEW preferable?**

A regular VIEW is preferable when data freshness is important, storage overhead should be avoided, or the query is inexpensive enough to execute on demand. A Materialized View stores computed results and must be refreshed when the underlying data changes.

**4. How does the recursive CTE prevent infinite recursion?**

The recursive query maintains a `path_ids` array and checks whether the next customer is already present in the path. In the test graph X → A → B → A, the repeated visit to A is rejected. The additional depth condition, `depth < 50`, provides a maximum recursion depth.

**5. Which trade-off matters most?**

For this customer-feature pipeline, I would prioritize query latency while preserving an acceptable freshness level. The Materialized View reduces repeated computation, while `REFRESH MATERIALIZED VIEW CONCURRENTLY` allows readers to continue accessing the existing view during refresh. The refresh frequency should reflect how often the source data changes and how current the features must be.
